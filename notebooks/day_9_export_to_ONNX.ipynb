```python
from ultralytics import YOLO
import os
import shutil
```

```python
from google.colab import drive
drive.mount("/content/drive")
```

```python
PROJECT_ROOT = "/content/drive/MyDrive/thermal project/Thermal-RGB-Integrated-Detection-with-Encoded-Noise-Tolerance"

MODEL_PATH = os.path.join(
    PROJECT_ROOT,
    "runs",
    "thermal_yolov8n_subset",
    "weights",
    "best.pt"
)

EXPORT_DIR = os.path.join(
    PROJECT_ROOT,
    "runs",
    "exports"
)

os.makedirs(EXPORT_DIR, exist_ok=True)

print("Model exists:", os.path.exists(MODEL_PATH))
```

```python
model = YOLO(MODEL_PATH)
```

```python
onnx_path = model.export(
    format="onnx",
    imgsz=640,
    dynamic=False,
    simplify=True,
    opset=12
)

print("Exported:", onnx_path)
```

```python
shutil.copy2(
    onnx_path,
    os.path.join(EXPORT_DIR, "best.onnx")
)

print("Saved to:", EXPORT_DIR)
```

```python
print("PT Model :",
      os.path.exists(MODEL_PATH))

print("ONNX Model:",
      os.path.exists(os.path.join(EXPORT_DIR, "best.onnx")))
```
